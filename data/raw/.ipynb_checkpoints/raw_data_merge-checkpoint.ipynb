```python
# Make copies so we do not modify our original DataFrames.
df_main = df.copy()
df_transcripts = df1.copy()

# Remove leading and trailing whitespace from URLs.
df_main["url"] = df_main["url"].astype("string").str.strip()
df_transcripts["url"] = df_transcripts["url"].astype("string").str.strip()

# Check the cleaned URLs.
print(repr(df_main["url"].iloc[0]))
print(repr(df_transcripts["url"].iloc[0]))
```

```python
# Count how many transcript rows repeat a URL.
print(
    "Duplicate transcript URLs:",
    df_transcripts["url"].duplicated().sum()
)

# Check whether repeated URLs have different transcript text.
conflicting_urls = (
    df_transcripts.groupby("url")["transcript"]
    .nunique(dropna=False)
)

conflicting_urls = conflicting_urls[conflicting_urls > 1]

print(
    "URLs with different transcript texts:",
    len(conflicting_urls)
)
```

```python
print("Duplicate URLs in main dataset:", df_main["url"].duplicated().sum())
```

```python
# Merge transcripts into the main TED Talk dataset.
# Remove duplicate transcript rows for the same URL before merging.
df_transcripts_unique = (
    df_transcripts[["url", "transcript"]]
    .drop_duplicates(subset="url", keep="first")
    .copy()
)

# Make sure URLs are cleaned consistently.
df_transcripts_unique["url"] = df_transcripts_unique["url"].astype("string").str.strip()

# Merge transcripts into the main TED Talk dataset.
df_merged = df_main.merge(
    df_transcripts_unique,
    on="url",
    how="left",
    validate="one_to_one",
    indicator=True
)

# Display the shape of the merged DataFrame.
print("Merged dataset shape:", df_merged.shape)

# Count successful matches and talks without transcripts.
print(df_merged["_merge"].value_counts())
```

```python
# Confirm that every main-dataset row is retained.
print("Main dataset rows:", len(df_main))
print("Merged dataset rows:", len(df_merged))

# Count talks with and without transcripts.
print("Talks with transcripts:", (df_merged["_merge"] == "both").sum())
print("Talks without transcripts:", (df_merged["_merge"] == "left_only").sum())

# Count missing transcript values.
print("Missing transcripts:", df_merged["transcript"].isna().sum())

# Inspect the combined data.
display(df_merged[["title", "main_speaker", "url", "transcript"]].head())
```

```python
# Keep the main talk details and transcript.
ted_talks = df_merged.drop(columns="_merge")

# Verify the final dataset.
print("Final dataset shape:", ted_talks.shape)

# Display selected columns.
display(
    ted_talks[
        ["title", "main_speaker", "tags", "description", "transcript"]
    ].head()
)
```

```python
# Save the combined dataset as a CSV file.
ted_talks.to_csv(
    "C:/Users/ajayr/Desktop/TED-TALK-RECOMMENDATION-SYSTEM/data/processed/ted_talks_merged.csv",
    index=False
)

print("Merged dataset saved successfully!")
```
