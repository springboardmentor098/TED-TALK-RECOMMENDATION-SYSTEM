```python
numeric_count_columns = ["comments", "duration", "languages", "num_speaker", "views"]
```

```python
display(df_clean[numeric_count_columns].lt(0).sum().rename("negative_value_count").to_frame())
```

```python
df_clean["film_date_datetime"] = pd.to_datetime(df_clean["film_date"], unit="s", errors="coerce", utc=True)
```

```python
df_clean["published_datetime"] = pd.to_datetime(df_clean["published_date"], unit="s", errors="coerce", utc=True)
```

## 6. Clean text fields

We preserve the original columns and make separate cleaned versions. Missing transcripts become empty strings **only in `transcript_clean`**; the original `transcript` column remains unchanged. Missing speaker occupations are labelled `unknown` in a separate cleaned column. Text is lowercased and repeated whitespace is collapsed. We also strip quote characters surrounding titles, where present.

```python
df_clean["title_clean"] = df_clean["title"].fillna("").astype("string").str.lower().str.replace(r"\s+", " ", regex=True).str.strip().str.strip(" \"'")
```

```python
df_clean["description_clean"] = df_clean["description"].fillna("").astype("string").str.lower().str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
df_clean["tags_clean"] = df_clean["tags"].fillna("").astype("string").str.lower().str.replace("[", "", regex=False).str.replace("]", "", regex=False).str.replace("'", "", regex=False).str.replace('"', "", regex=False).str.replace(",", " ", regex=False).str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
df_clean["transcript_clean"] = df_clean["transcript"].fillna("").astype("string").str.lower().str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
df_clean["speaker_occupation_clean"] = df_clean["speaker_occupation"].fillna("unknown").astype("string").str.lower().str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
display(df_clean[["title", "title_clean", "tags", "tags_clean", "speaker_occupation", "speaker_occupation_clean"]].head())
```

### Validate missing values after text cleaning

```python
clean_text_columns = ["title_clean", "description_clean", "tags_clean", "transcript_clean", "speaker_occupation_clean"]
```

## 7. Create the prepared text field

`content_text` combines the title, description, tags, and transcript into one text field per talk. This field is prepared for the separate model notebook. No vectorization or recommendation logic is performed here.

```python
df_clean["content_text"] = df_clean[["title_clean", "description_clean", "tags_clean", "transcript_clean"]].fillna("").agg(" ".join, axis=1).str.replace(r"\s+", " ", regex=True).str.strip()
```

```python
df_clean["content_length"] = df_clean["content_text"].str.len()
```

```python
display(df_clean[["title", "content_length", "content_text"]].head())
```

```python
print("Empty content records:", int(df_clean["content_text"].str.strip().eq("").sum()))
```

```python
print("Missing values in content_text:", int(df_clean["content_text"].isna().sum()))
```

```python
display(df_clean.nsmallest(5, "content_length")[["title", "content_length", "content_text"]])
```

## 8. Detect and inspect numeric outliers

We use the **IQR rule**: values below `Q1 − 1.5 × IQR` or above `Q3 + 1.5 × IQR` are flagged as potential outliers. The date timestamp columns are excluded because they represent dates, not measurable quantities.

**Important:** flagging is not the same as proving a value is wrong. Views, comments, and duration can be naturally skewed. For this text-based recommender preparation, we will not automatically delete or cap these records. We preserve them and save an outlier report for review.

```python
outlier_columns = ["comments", "duration", "languages", "num_speaker", "views"]
```

```python
q1 = df_clean[outlier_columns].quantile(0.25)
```

```python
q3 = df_clean[outlier_columns].quantile(0.75)
```

```python
iqr = q3 - q1
```

```python
lower_bound = q1 - 1.5 * iqr
```

```python
upper_bound = q3 + 1.5 * iqr
```

```python
outlier_summary = pd.DataFrame({"column": outlier_columns, "q1": q1.to_numpy(), "q3": q3.to_numpy(), "iqr": iqr.to_numpy(), "lower_bound": lower_bound.to_numpy(), "upper_bound": upper_bound.to_numpy()})
```

```python
outlier_summary["outlier_count"] = [int(((df_clean[column] < lower_bound[column]) | (df_clean[column] > upper_bound[column])).sum()) for column in outlier_columns]
```

```python
outlier_summary["outlier_percent"] = (outlier_summary["outlier_count"] / len(df_clean) * 100).round(2)
```

```python
display(outlier_summary)
```

### Visual outlier inspection

The first boxplot uses a logarithmic vertical scale because comments, duration, and views can have wide, right-skewed distributions. This changes only how the plot is displayed; it does not alter the underlying values.
