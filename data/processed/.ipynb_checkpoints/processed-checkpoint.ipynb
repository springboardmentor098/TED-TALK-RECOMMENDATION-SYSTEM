```python

# Step 1: Create a copy of the dataset
df_clean = df.copy()

# Step 2: Clean the tags column
df_clean["tags_clean"] = (
    df_clean["tags"]
    .fillna("")
    .astype(str)
    .str.replace(r"[\[\]']", "", regex=True)
    .str.replace(",", " ", regex=False)
    .str.lower()
    .str.replace(r"\s+", " ", regex=True)
    .str.strip()
)

# Step 3: Clean the other text columns
text_columns = ["title", "description", "transcript"]

for column in text_columns:
    df_clean[column + "_clean"] = (
        df_clean[column]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

# Step 4: Check the cleaned columns
display(
    df_clean[
        [
            "title",
            "title_clean",
            "tags",
            "tags_clean",
            "transcript_clean"
        ]
    ].head()
)
```

```python

print("Original rows:", len(df))
print("Cleaned rows:", len(df_clean))

print(
    "Missing original transcripts:",
    df_clean["transcript"].isna().sum()
)

print(
    "Empty cleaned transcripts:",
    df_clean["transcript_clean"].eq("").sum()
)

print(
    "Missing values in cleaned title:",
    df_clean["title_clean"].isna().sum()
)
```

```python
# Combine the cleaned text fields into one field per TED Talk.
df_clean["content_text"] = (
    df_clean["title_clean"] + " " +
    df_clean["description_clean"] + " " +
    df_clean["tags_clean"] + " " +
    df_clean["transcript_clean"]
).str.strip()

# Check the result.
display(
    df_clean[
        ["title", "content_text"]
    ].head()
)

print("Total rows:", len(df_clean))
print(
    "Empty content records:",
    df_clean["content_text"].eq("").sum()
)
```

```python
# 1. Check duplicate URLs
print("Duplicate URLs:", df_clean["url"].duplicated().sum())

# 2. Check duplicate titles
print("Duplicate titles:", df_clean["title_clean"].duplicated().sum())

# 3. Calculate the length of each combined text
df_clean["content_length"] = df_clean["content_text"].str.len()

# 4. Inspect text length statistics
print(df_clean["content_length"].describe())

# 5. Display the 5 shortest records
display(
    df_clean.nsmallest(5, "content_length")[
        ["title", "content_length", "content_text"]
    ]
)
```

```python

# Count missing values in every original column
missing_counts = df.isna().sum()

# Display only columns that contain missing values
print(missing_counts[missing_counts > 0])
```

```python

# Check missing values in the cleaned text columns
clean_text_columns = [
    "title_clean",
    "description_clean",
    "tags_clean",
    "transcript_clean",
    "content_text"
]

print("Missing values in cleaned columns:")
print(df_clean[clean_text_columns].isna().sum())

print(
    "\nEmpty content records:",
    df_clean["content_text"].str.strip().eq("").sum()
)
```
