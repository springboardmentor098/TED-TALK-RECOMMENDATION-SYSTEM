```python
# Activity 2: Design User Interaction Dataset

import pandas as pd
import os
```

```python
# Load the existing cleaned TED Talks dataset

df = pd.read_csv("../data/processed/ted_talks_cleaned.csv")

print("Dataset shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
# Activity 2: Create Simulated User Interaction Dataset

talk_ids = df['talk_id'].sample(n=5, random_state=42).tolist()

user_interactions = pd.DataFrame({
    'user_id': ['U001', 'U001', 'U001',
                'U002', 'U002', 'U002',
                'U003', 'U003', 'U003'],

    'talk_id': [talk_ids[0], talk_ids[1], talk_ids[2],
                talk_ids[0], talk_ids[2], talk_ids[3],
                talk_ids[1], talk_ids[3], talk_ids[4]],

    'rating': [5, 4, 3,
               4, 5, 2,
               5, 3, 4]
})

os.makedirs("../data/interaction", exist_ok=True)

user_interactions.to_csv(
    "../data/interaction/user_interactions.csv",
    index=False
)

print("Simulated user interaction dataset created successfully!")
```

```python
# Activity 3: Create User-Item Interaction Matrix

interaction_matrix = user_interactions.pivot(
    index='user_id',
    columns='talk_id',
    values='rating'
)

interaction_matrix
```

```python
# Activity 4: Basic Interaction Data Analysis

print("Number of unique users:", user_interactions['user_id'].nunique())

print("Number of unique TED Talks:", user_interactions['talk_id'].nunique())
```

```python
# Find TED Talks with the highest average ratings

talk_ratings = user_interactions.groupby('talk_id')['rating'].mean()

print(talk_ratings.sort_values(ascending=False))
```

```python
# Find users who rated the most TED Talks

user_ratings = user_interactions.groupby('user_id')['talk_id'].count()

print(user_ratings.sort_values(ascending=False))
```

```python
# Find TED Talks with the most interactions

talk_interactions = user_interactions.groupby('talk_id')['user_id'].count()

print(talk_interactions.sort_values(ascending=False))
```

### What does a missing rating indicate?

A missing rating (NaN) indicates that a user has not rated or interacted with a particular TED Talk.

It does not necessarily mean that the user disliked the talk. It may simply mean that the user has not watched or rated it.

## Activity 5: Understanding Collaborative Filtering

Collaborative filtering is a recommendation technique that suggests items to users based on the preferences and interests of other users.

It identifies users who have similar interests by comparing their ratings and interactions.

For example:

If User A and User B have rated several TED Talks similarly, the system may identify that they have similar interests.

If User A liked a TED Talk that User B has not watched, the system may recommend that TED Talk to User B.

### How Collaborative Filtering Helps

- Identifies users with similar interests.
- Uses previous ratings and interactions.
- Recommends TED Talks based on user preferences.
- Provides personalized recommendations to different users.
