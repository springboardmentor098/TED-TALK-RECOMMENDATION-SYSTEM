# Markdown -> PDF Converter

Converts every `.md` report in `../reports` into a PDF in `../reports/pdf` using the [`markdown-pdf`](https://pypi.org/project/markdown-pdf/) package.

- No `pandoc` or LaTeX installation is required.
- Run this notebook from the `notebooks/` folder, like the other notebooks.
- Re-run after adding or editing a report.
- Existing PDFs with the same filename are overwritten.


```python
%pip install markdown-pdf
```

```python
from pathlib import Path

from markdown_pdf import MarkdownPdf, Section

REPORTS_DIR = Path("../reports")
PDF_DIR = REPORTS_DIR / "pdf"
PDF_DIR.mkdir(parents=True, exist_ok=True)

md_files = sorted(REPORTS_DIR.glob("*.md"))

if not md_files:
    print(f"No .md files found in {REPORTS_DIR.resolve()}")
else:
    print(f"Found {len(md_files)} Markdown file(s) in {REPORTS_DIR.resolve()}\n")

    converted = 0
    failed = []

    for md_path in md_files:
        pdf_path = PDF_DIR / f"{md_path.stem}.pdf"
        try:
            md_text = md_path.read_text(encoding="utf-8")
            pdf = MarkdownPdf()
            pdf.add_section(Section(md_text))
            pdf.save(str(pdf_path))
            converted += 1
            print(f"  OK   {md_path.name} -> {pdf_path.relative_to(REPORTS_DIR)}")
        except Exception as exc:
            failed.append((md_path.name, str(exc)))
            print(f"  FAIL {md_path.name}: {exc}")

    print(f"\nConverted {converted}/{len(md_files)} report(s) to {PDF_DIR.resolve()}")

    if failed:
        print("\nFailures:")
        for name, error in failed:
            print(f"  - {name}: {error}")
```

## Notes

- Source reports stay in `reports/*.md`.
- PDFs are written to `reports/pdf/*.pdf`.
- To add a report, drop a `.md` file into `reports/` and re-run the conversion cell.
- The first cell installs `markdown-pdf` into the active kernel environment.
- The converter skips subdirectories, so files already in `reports/pdf/` are not processed again.
