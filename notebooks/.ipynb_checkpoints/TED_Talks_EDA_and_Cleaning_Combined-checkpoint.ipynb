# TED Talks Project: EDA, Data Quality, and Data Cleaning

**Beginner-friendly combined notebook**

This notebook does the complete workflow in one place:
1. Finds the dataset file(s) in `data/raw/`.
2. Explores the data using summary statistics and missing-value/duplicate checks.
3. Creates visualizations for text length, tags, numeric columns, and missing values where possible.
4. Detects numeric outliers and replaces them using a method you can choose.
5. Cleans text and creates `combined_text` for the recommendation system.
6. Saves the cleaned dataset and reports without changing the original files.

**Before running:** Put your TED Talks dataset file(s) in `data/raw/`. This notebook supports CSV, TSV, JSON, XLSX, and XLS files. If there are multiple files in the folder, it will try to load and combine them. Keep only files belonging to this dataset in that folder.

## 1. Import the libraries

Pandas loads and analyzes tables. Matplotlib and Seaborn make charts. NumPy helps with missing values.

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

## 2. Find and load the dataset files

The notebook checks whether you opened it from the project folder or from the `notebooks/` folder. It reads all supported files in `data/raw/` and combines them into one table.

```python
# Find the project folders automatically.
if Path("data/raw").exists():
    RAW_FOLDER = Path("data/raw")
elif Path("../data/raw").exists():
    RAW_FOLDER = Path("../data/raw")
else:
    raise FileNotFoundError("Create a data/raw/ folder and place your dataset file(s) inside it.")

PROJECT_FOLDER = RAW_FOLDER.parent.parent
PROCESSED_FOLDER = PROJECT_FOLDER / "data" / "processed"
REPORTS_FOLDER = PROJECT_FOLDER / "reports"
PROCESSED_FOLDER.mkdir(parents=True, exist_ok=True)
REPORTS_FOLDER.mkdir(parents=True, exist_ok=True)

# Find supported files.
data_files = []
for pattern in ["*.csv", "*.tsv", "*.json", "*.xlsx", "*.xls"]:
    data_files.extend(RAW_FOLDER.glob(pattern))
data_files = sorted(data_files)

if len(data_files) == 0:
    raise FileNotFoundError(f"No CSV, TSV, JSON, Excel files found in {RAW_FOLDER}")

# Read each file. Files that cannot be read are reported below.
data_frames = []
for file in data_files:
    try:
        if file.suffix.lower() == ".csv":
            part = pd.read_csv(file, encoding_errors="replace")
        elif file.suffix.lower() == ".tsv":
            part = pd.read_csv(file, sep="\t", encoding_errors="replace")
        elif file.suffix.lower() == ".json":
            try:
                part = pd.read_json(file)
            except ValueError:
                part = pd.read_json(file, lines=True)
        else:
            part = pd.read_excel(file)

        data_frames.append(part)
        print(f"Loaded: {file.name} | rows: {part.shape[0]} | columns: {part.shape[1]}")
    except Exception as error:
        print(f"Could not load {file.name}: {error}")

if len(data_frames) == 0:
    raise ValueError("None of the files could be loaded. Check the file format and try again.")

df = pd.concat(data_frames, ignore_index=True, sort=False)

# Treat blank or whitespace-only cells as missing values for analysis.
df = df.replace(r"^\s*$", np.nan, regex=True)

print("\nCombined dataset loaded successfully.")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
display(df.head())
```

## 3. Dataset summary

Check the size of the dataset, column names, data types, descriptive statistics, and number of unique talks. The notebook looks for common names such as `talk_id`, `url`, or `title`; it does not assume every column exists.

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names and data types:")
display(pd.DataFrame({"column": df.columns, "data_type": df.dtypes.astype(str).values}))

print("\nDescriptive statistics:")
display(df.describe(include="all").T)

# Find common TED Talk columns, when available.
title_col = None
transcript_col = None
description_col = None
tags_col = None
url_col = None
id_col = None
duration_col = None
views_col = None

for column in df.columns:
    name = str(column).lower().strip().replace(" ", "_")
    if title_col is None and "title" in name:
        title_col = column
    if transcript_col is None and "transcript" in name:
        transcript_col = column
    if description_col is None and ("description" in name or "summary" in name):
        description_col = column
    if tags_col is None and ("tag" in name or "topic" in name):
        tags_col = column
    if url_col is None and ("url" in name or "link" in name):
        url_col = column
    if id_col is None and (name == "id" or name.endswith("_id")):
        id_col = column
    if duration_col is None and ("duration" in name or "length" in name):
        duration_col = column
    if views_col is None and "view" in name:
        views_col = column

print("\nDetected columns:")
print("Title:", title_col)
print("Transcript:", transcript_col)
print("Description:", description_col)
print("Tags/topics:", tags_col)
print("URL/link:", url_col)
print("Talk ID:", id_col)
print("Duration:", duration_col)
print("Views:", views_col)

if id_col is not None:
    print("\nUnique talk IDs:", df[id_col].nunique())
elif url_col is not None:
    print("\nUnique talk URLs:", df[url_col].nunique())
elif title_col is not None:
    print("\nUnique talk titles:", df[title_col].nunique())
else:
    print("\nNo common talk ID, URL, or title column was found.")
```

## 4. Missing values and duplicates

Missing data is not automatically deleted. First, inspect how much is missing and which columns are affected. Exact duplicate rows are counted here; duplicate IDs and URLs are also checked when those columns exist.

```python
missing_report = pd.DataFrame({
    "column": df.columns,
    "missing_count": df.isna().sum().values,
    "missing_percent": (df.isna().mean().values * 100).round(2)
}).sort_values("missing_percent", ascending=False)

display(missing_report)

exact_duplicates = df.duplicated().sum()
print("Exact duplicate rows:", exact_duplicates)

if id_col is not None:
    print("Repeated non-empty talk IDs:", df[id_col].dropna().duplicated().sum())
if url_col is not None:
    print("Repeated non-empty URLs:", df[url_col].dropna().duplicated().sum())

missing_report.to_csv(REPORTS_FOLDER / "missing_values_report.csv", index=False)
```

## 5. Basic text analysis

For available title, description, and transcript fields, calculate the number of words and find empty or unusually short entries. A missing transcript does not automatically mean a talk must be removed; title, tags, and description may still support recommendations.

```python
text_columns = [column for column in [title_col, description_col, transcript_col] if column is not None]
text_length_results = []

for column in text_columns:
    word_counts = df[column].fillna("").astype(str).str.split().str.len()
    text_length_results.append({
        "column": column,
        "empty_entries": int((word_counts == 0).sum()),
        "minimum_words": int(word_counts.min()) if len(word_counts) else 0,
        "median_words": float(word_counts.median()) if len(word_counts) else 0,
        "average_words": round(float(word_counts.mean()), 2) if len(word_counts) else 0,
        "maximum_words": int(word_counts.max()) if len(word_counts) else 0
    })
    print(f"{column}: {int((word_counts == 0).sum())} empty entries")

if text_length_results:
    text_length_report = pd.DataFrame(text_length_results)
    display(text_length_report)
    text_length_report.to_csv(REPORTS_FOLDER / "text_length_report.csv", index=False)
else:
    print("No title, description, or transcript column was detected. Text analysis was skipped.")
```

## 6. Tag and topic analysis

Tags may be stored as comma-separated text or as a list-like value. The simple processing below removes list brackets and quotes, then splits tags on commas, semicolons, or vertical bars. It counts the most common tags without changing the original dataset.

```python
tag_counts = pd.Series(dtype="int64")

if tags_col is not None:
    tag_text = df[tags_col].dropna().astype(str).str.lower()
    tag_text = tag_text.str.replace("[", "", regex=False).str.replace("]", "", regex=False)
    tag_text = tag_text.str.replace("'", "", regex=False).str.replace('"', "", regex=False)
    tag_counts = tag_text.str.split(r"[,;|]").explode().str.strip()
    tag_counts = tag_counts[tag_counts != ""].value_counts().head(15)

    print("Top tags/topics:")
    display(tag_counts.rename_axis("tag").reset_index(name="count"))
    tag_counts.rename_axis("tag").reset_index(name="count").to_csv(REPORTS_FOLDER / "tag_frequency_report.csv", index=False)
else:
    print("No tags/topic column was found. Tag analysis was skipped.")
```

## 7. Visualizations

The following cells create charts when the relevant fields exist. These charts help show missing data, text-length patterns, popular tags, and numeric distributions.

```python
# Chart 1: Missing-value percentages
missing_to_plot = missing_report[missing_report["missing_count"] > 0].head(15)
plt.figure(figsize=(10, 5))
if len(missing_to_plot) > 0:
    sns.barplot(data=missing_to_plot, x="missing_percent", y="column", color="steelblue")
    plt.title("Columns with the Most Missing Values")
    plt.xlabel("Missing values (%)")
    plt.ylabel("Column")
else:
    plt.text(0.5, 0.5, "No missing values detected", ha="center", va="center")
    plt.title("Missing-Value Check")
    plt.axis("off")
plt.tight_layout()
plt.savefig(REPORTS_FOLDER / "01_missing_values.png", dpi=150, bbox_inches="tight")
plt.show()
```

```python
# Chart 2: Transcript word-count distribution (or description if transcript is absent)
text_column_for_chart = transcript_col if transcript_col is not None else description_col

if text_column_for_chart is not None:
    word_counts = df[text_column_for_chart].fillna("").astype(str).str.split().str.len()
    plt.figure(figsize=(9, 5))
    sns.histplot(word_counts, bins=30, kde=True, color="steelblue")
    plt.title(f"Word-Count Distribution: {text_column_for_chart}")
    plt.xlabel("Number of words")
    plt.ylabel("Number of talks")
    plt.tight_layout()
    plt.savefig(REPORTS_FOLDER / "02_text_length_distribution.png", dpi=150, bbox_inches="tight")
    plt.show()
else:
    print("No transcript or description column was found, so the text-length chart was skipped.")
```

```python
# Chart 3: Top tags/topics
if len(tag_counts) > 0:
    plt.figure(figsize=(10, 6))
    sns.barplot(x=tag_counts.values, y=tag_counts.index, color="steelblue")
    plt.title("Most Frequent TED Talks Tags/Topics")
    plt.xlabel("Number of occurrences")
    plt.ylabel("Tag/topic")
    plt.tight_layout()
    plt.savefig(REPORTS_FOLDER / "03_top_tags.png", dpi=150, bbox_inches="tight")
    plt.show()
else:
    print("Tag chart skipped because a usable tags/topic column was not found.")
```

```python
# Chart 4: Distributions for numeric fields such as views or duration
numeric_chart_columns = []
for column in [views_col, duration_col]:
    if column is not None and pd.api.types.is_numeric_dtype(df[column]):
        numeric_chart_columns.append(column)

for column in numeric_chart_columns:
    plt.figure(figsize=(9, 5))
    sns.histplot(df[column].dropna(), bins=30, kde=True, color="steelblue")
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Number of talks")
    plt.tight_layout()
    plt.savefig(REPORTS_FOLDER / f"distribution_{str(column).replace(' ', '_')}.png", dpi=150, bbox_inches="tight")
    plt.show()

if not numeric_chart_columns:
    print("No numeric views/duration column was found. Numeric distribution charts were skipped.")
```

## 8. Detect numeric outliers

This notebook uses the **IQR method**. It marks values below `Q1 - 1.5 × IQR` or above `Q3 + 1.5 × IQR` as possible outliers. An outlier is not always an error; unusually high views may be real.

The report below shows the number of possible outliers in each numeric column. ID columns are skipped so identifiers are not treated as measurements.

```python
outlier_results = []
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if "id" in str(column).lower():
        continue

    q1 = df[column].quantile(0.25)
    q3 = df[column].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    outlier_mask = (df[column] < lower_limit) | (df[column] > upper_limit)

    outlier_results.append({
        "column": column,
        "outlier_count": int(outlier_mask.sum()),
        "lower_limit": lower_limit,
        "upper_limit": upper_limit
    })

outlier_report = pd.DataFrame(outlier_results)
if len(outlier_report) > 0:
    display(outlier_report)
    outlier_report.to_csv(REPORTS_FOLDER / "outlier_report.csv", index=False)
else:
    print("No suitable numeric measurement columns were found for outlier analysis.")
```

These findings are generated from the dataset you loaded. Read them and add your own explanation in the final report if needed.

```python
print("FINDINGS")
print("1. Dataset size:", df.shape[0], "rows and", df.shape[1], "columns.")
print("2. Exact duplicate rows:", int(df.duplicated().sum()))

if len(missing_report) > 0:
    most_missing = missing_report.iloc[0]
    print(f"3. Column with the highest missing percentage: {most_missing['column']} ({most_missing['missing_percent']}%).")
else:
    print("3. No columns were available for missing-value analysis.")

if text_columns:
    print("4. Text fields available for analysis:", ", ".join(map(str, text_columns)))
else:
    print("4. No common text fields were detected.")

if len(tag_counts) > 0:
    print(f"5. Most frequent tag/topic: '{tag_counts.index[0]}' ({int(tag_counts.iloc[0])} occurrences).")
else:
    print("5. Tag frequency could not be calculated because no tags/topic field was detected.")

if len(outlier_report) > 0:
    total_outliers = int(outlier_report["outlier_count"].sum())
    print("6. Possible outliers found across numeric measurement columns:", total_outliers)
else:
    print("6. No numeric measurement columns were available for outlier analysis.")
```

## 10. Clean exact duplicate rows and text fields

The original dataset in `data/raw/` is never overwritten. This step works on a copy.

- Exact duplicate rows are removed.
- Missing transcripts are kept; other fields may still be useful for recommendations.
- Text fields get new columns ending in `_clean`, so original text remains available for comparison.
- Repeated IDs or URLs were reported above. Exact rows are removed automatically; conflicting records sharing an ID/URL are left for review rather than risking accidental data loss.

```python
clean_df = df.copy()
rows_before = len(clean_df)
exact_duplicates_before = int(clean_df.duplicated().sum())

clean_df = clean_df.drop_duplicates().copy()
rows_after_exact_deduplication = len(clean_df)

print("Rows before duplicate removal:", rows_before)
print("Exact duplicate rows found:", exact_duplicates_before)
print("Rows after exact duplicate removal:", rows_after_exact_deduplication)

# Clean available text fields in new columns; preserve original columns.
text_columns_to_clean = [column for column in [title_col, tags_col, description_col, transcript_col] if column is not None]
cleaned_text_columns = []

for column in text_columns_to_clean:
    cleaned_column = str(column) + "_clean"
    clean_df[cleaned_column] = clean_df[column].fillna("").astype(str).str.lower().str.strip()
    clean_df[cleaned_column] = clean_df[cleaned_column].str.replace(r"\s+", " ", regex=True)
    cleaned_text_columns.append(cleaned_column)

print("Text fields cleaned:", text_columns_to_clean if text_columns_to_clean else "None found")
```

## 11. Create `combined_text`

The recommendation system will use this field later. It combines whichever of title, tags, description, and transcript were found. Missing text becomes an empty string, so talks without transcripts are not automatically removed.

```python
clean_df["combined_text"] = ""

for column in cleaned_text_columns:
    clean_df["combined_text"] = clean_df["combined_text"] + " " + clean_df[column]

clean_df["combined_text"] = clean_df["combined_text"].str.replace(r"\s+", " ", regex=True).str.strip()

print("Example of original and cleaned text:")
if title_col is not None:
    display(clean_df[[title_col, str(title_col) + "_clean", "combined_text"]].head())
else:
    display(clean_df[["combined_text"]].head())
```

## 12. Replace detected outliers

Choose one method below:

- `"cap"`: replace extreme values with the IQR lower/upper limits.
- `"median"`: replace extreme values with the column median.
- `"mean"`: replace extreme values with the column mean.

The default is `"cap"`. Adjusted values are saved in new columns ending in `_clean`; the original numeric columns are preserved. This also makes it easy to compare original and adjusted values.

```python
OUTLIER_METHOD = "cap"  # Change to "median" or "mean" if your mentor requests it.

if OUTLIER_METHOD not in ["cap", "median", "mean"]:
    raise ValueError("OUTLIER_METHOD must be 'cap', 'median', or 'mean'.")

cleaning_outlier_results = []
numeric_columns = clean_df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    if "id" in str(column).lower():
        continue

    q1 = clean_df[column].quantile(0.25)
    q3 = clean_df[column].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    outlier_mask = (clean_df[column] < lower_limit) | (clean_df[column] > upper_limit)
    outlier_count = int(outlier_mask.sum())

    cleaned_column = str(column) + "_clean"
    clean_df[cleaned_column] = clean_df[column]

    if outlier_count > 0:
        if OUTLIER_METHOD == "cap":
            clean_df[cleaned_column] = clean_df[column].clip(lower=lower_limit, upper=upper_limit)
        elif OUTLIER_METHOD == "median":
            clean_df.loc[outlier_mask, cleaned_column] = clean_df[column].median()
        else:
            clean_df.loc[outlier_mask, cleaned_column] = clean_df[column].mean()

    cleaning_outlier_results.append({
        "column": column,
        "outliers_found": outlier_count,
        "replacement_method": OUTLIER_METHOD,
        "cleaned_column": cleaned_column
    })

cleaning_outlier_report = pd.DataFrame(cleaning_outlier_results)
if len(cleaning_outlier_report) > 0:
    display(cleaning_outlier_report)
    cleaning_outlier_report.to_csv(REPORTS_FOLDER / "outlier_replacement_report.csv", index=False)
else:
    print("No numeric measurement columns were available for outlier replacement.")
```

## 13. Save the cleaned dataset and reports

The cleaned dataset is saved to `data/processed/ted_talks_cleaned.csv`. The original input files in `data/raw/` remain unchanged. The notebook also saves summary reports and chart images in `reports/`.

```python
cleaned_file = PROCESSED_FOLDER / "ted_talks_cleaned.csv"
clean_df.to_csv(cleaned_file, index=False)

cleaning_summary = pd.DataFrame({
    "measure": [
        "rows_before_cleaning",
        "columns_before_cleaning",
        "exact_duplicate_rows_found",
        "rows_after_exact_duplicate_removal",
        "rows_in_saved_dataset",
        "columns_in_saved_dataset"
    ],
    "value": [
        rows_before,
        df.shape[1],
        exact_duplicates_before,
        rows_after_exact_deduplication,
        clean_df.shape[0],
        clean_df.shape[1]
    ]
})
cleaning_summary.to_csv(REPORTS_FOLDER / "cleaning_summary.csv", index=False)

# Save a simple data-quality summary.
quality_summary = pd.DataFrame({
    "measure": ["rows", "columns", "exact_duplicate_rows", "missing_cells"],
    "value": [df.shape[0], df.shape[1], int(df.duplicated().sum()), int(df.isna().sum().sum())]
})
quality_summary.to_csv(REPORTS_FOLDER / "data_quality_summary.csv", index=False)

display(cleaning_summary)
print("\nCleaned dataset saved to:", cleaned_file)
print("Reports and charts saved to:", REPORTS_FOLDER)
print("Original files in data/raw/ were not changed.")
```

## 14. Data-cleaning plan and interpretation

Use the generated reports to explain your choices:

- **Missing titles:** review these talks; keep them if other usable information exists.
- **Missing transcripts:** keep the talk if its title, tags, or description can still support recommendations.
- **Missing tags:** leave them blank unless another reliable source can supply them.
- **Exact duplicates:** remove them because the full records are identical.
- **Repeated IDs or URLs with conflicting fields:** inspect manually before deleting any record.
- **Text fields:** use the `_clean` columns for lowercase and whitespace-normalized text; originals remain for comparison.
- **Outliers:** review the outlier reports before deciding if capping, median, or mean replacement is appropriate. Some extreme views or durations may be real.
- **Empty `combined_text`:** review these rows because there is no text available for content-based recommendation.
