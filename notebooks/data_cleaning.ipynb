### Activity 4: Clean Text Fields 

Clean the available text fields, such as title, tags, description, and transcript. 

Your preprocessing should include: 

 Converting text to lowercase where appropriate. 

 Removing unnecessary leading, trailing, and repeated whitespace. 

 Handling empty strings and missing text consistently. 

 Applying appropriate punctuation handling without damaging meaningful text. 

```python
import ast # Safely parses a string representation of a Python list, such as the topics values in your dataset.
import html # Converts HTML entities such as &amp; into their readable characters.
import re # Performs pattern-based text cleaning using regular expressions.
```

```python
df_clean = df.copy()
```

### We want to standardize text so that irrelevant differences in capitalization, whitespace, or HTML encoding don't unnecessarily affect text analysis.

```python
def clean_text(value, lowercase=True):
    """Clean text while preserving meaningful punctuation."""

    if pd.isna(value):
        return ""

    text = str(value)

    # Decode HTML entities such as &amp; and &quot;
    text = html.unescape(text)

    # Remove HTML tags, if present
    text = re.sub(r"<[^>]+>", " ", text)

    # Remove leading/trailing and repeated whitespace
    text = re.sub(r"\s+", " ", text).strip()

    # Convert to lowercase for consistent text processing
    if lowercase:
        text = text.lower()

    return text
```

### Each topics value is stored as a string containing a list representation, not as an actual Python list. That's why we need the separate clean_topics() function.

```python
def clean_topics(value):
    """Convert list-like topics into readable, cleaned text."""

    if pd.isna(value):
        return ""

    # Parse the string -> Python converts the string into an actual list
    if isinstance(value, str):
        try:
            parsed = ast.literal_eval(value) 
            if isinstance(parsed, (list, tuple)):
                value = parsed
        except (ValueError, SyntaxError):
            pass

    ''' Convert individual topics into a single text string Because a recommender system needs the topics to be 
    represented as usable text. '''
    if isinstance(value, (list, tuple)):
        value = " ".join(str(topic) for topic in value)

    return clean_text(value)
```

```python
# Preserve originals and create cleaned text columns
df_clean["title_clean"] = df_clean["title"].apply(clean_text)
df_clean["speaker_1_clean"] = df_clean["speaker_1"].apply(clean_text)
df_clean["topics_clean"] = df_clean["topics"].apply(clean_topics)
df_clean["description_clean"] = df_clean["description"].apply(clean_text)
df_clean["transcript_clean"] = df_clean["transcript"].apply(clean_text)
```

### Activity 5: Create a Combined Text Field 
Create a new column called combined_text by combining available fields such as title, tags, 
description, and transcript. 

### Now we want one column that represents the content of each TED Talk.

```python
# Combine cleaned fields into one text representation
text_columns = [
    "title_clean",
    "topics_clean",
    "description_clean",
    "transcript_clean"
]
```

```python
df_clean["combined_text"] = (
    df_clean[text_columns]
    .fillna("")
    .agg(" ".join, axis=1)
    .str.replace(r"\s+", " ", regex=True) # Normalizes repeated whitespace in the combined string.
    .str.strip()
)
```
