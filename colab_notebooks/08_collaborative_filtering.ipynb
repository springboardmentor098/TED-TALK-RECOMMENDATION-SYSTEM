```python

# ============================================================
# TASK 8: COLLABORATIVE FILTERING RECOMMENDATION MODEL
# Notebook: 08_collaborative_filtering.ipynb
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics.pairwise import cosine_similarity

# ------------------------------------------------------------
# 1. LOAD DATASETS
# ------------------------------------------------------------

INTERACTION_PATH = (
    "/content/user_interactions.csv"
)

TALKS_PATH = (
    "/content/ted_talks_cleaned.csv"
)

if (
    not Path(INTERACTION_PATH).exists()
    or not Path(TALKS_PATH).exists()
):

    from google.colab import files

    print("Upload user_interactions.csv and ted_talks_cleaned.csv")
    files.upload()

interactions = pd.read_csv(
    INTERACTION_PATH,
    dtype={"user_id": str}
)

talks = pd.read_csv(TALKS_PATH)

print("Interaction dataset:", interactions.shape)
print("TED Talks dataset:", talks.shape)

# ------------------------------------------------------------
# 2. STANDARDIZE IDENTIFIERS
# ------------------------------------------------------------

def normalize_id(value):

    if pd.isna(value):
        return ""

    value = str(value).strip()

    if value.endswith(".0"):
        value = value[:-2]

    return value


interactions["user_id"] = (
    interactions["user_id"].astype(str).str.strip()
)

interactions["talk_id"] = (
    interactions["talk_id"].apply(normalize_id)
)

talks["talk_id"] = (
    talks["talk_id"].apply(normalize_id)
)

interactions["rating"] = pd.to_numeric(
    interactions["rating"],
    errors="coerce"
)

interactions = interactions.dropna(
    subset=["user_id", "talk_id", "rating"]
)

interactions = interactions[
    interactions["talk_id"] != ""
].copy()

# Remove repeated user-talk interactions
interactions = interactions.drop_duplicates(
    subset=["user_id", "talk_id"],
    keep="last"
)

print("\nInteraction records:", len(interactions))

# ------------------------------------------------------------
# 3. CREATE USER-ITEM INTERACTION MATRIX
# ------------------------------------------------------------

user_item_matrix = interactions.pivot_table(

    index="user_id",

    columns="talk_id",

    values="rating",

    aggfunc="mean"

)

print("\nUSER-ITEM MATRIX")
print("Dimensions:", user_item_matrix.shape)

display(user_item_matrix.head(10))

# ------------------------------------------------------------
# 4. FILL MISSING VALUES WITH ZERO
# ------------------------------------------------------------

# Zero means no recorded interaction.
# It is NOT an actual user rating.

user_item_filled = user_item_matrix.fillna(0)

print("\nUser-item matrix after filling missing values:")

display(user_item_filled.head(10))

# ------------------------------------------------------------
# 5. CALCULATE USER SIMILARITY
# ------------------------------------------------------------

user_similarity = cosine_similarity(
    user_item_filled
)

user_similarity_df = pd.DataFrame(

    user_similarity,

    index=user_item_filled.index,

    columns=user_item_filled.index

)

print("\nUSER SIMILARITY MATRIX")

display(user_similarity_df.head(10))

# ------------------------------------------------------------
# 6. VISUALIZE USER SIMILARITY
# ------------------------------------------------------------

plt.figure(figsize=(10, 8))

sns.heatmap(
    user_similarity_df,
    cmap="Blues",
    annot=False
)

plt.title("User Similarity Matrix")

plt.xlabel("Users")

plt.ylabel("Users")

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# 7. IDENTIFY SIMILAR USERS
# ------------------------------------------------------------

def find_similar_users(user_id, top_n=5):

    user_id = str(user_id)

    if user_id not in user_similarity_df.index:

        print("User ID not found:", user_id)

        return pd.DataFrame()

    scores = user_similarity_df.loc[user_id].drop(
        user_id
    )

    scores = scores.sort_values(
        ascending=False
    )

    results = scores.head(top_n).reset_index()

    results.columns = [
        "Similar User ID",
        "Cosine Similarity"
    ]

    return results


# ------------------------------------------------------------
# 8. CREATE TALK TITLE LOOKUP
# ------------------------------------------------------------

talk_lookup = (
    talks.drop_duplicates("talk_id")
    .set_index("talk_id")
)

# ------------------------------------------------------------
# 9. RECOMMENDATION FUNCTION
# ------------------------------------------------------------

def recommend_for_user(
    user_id,
    top_n=5,
    n_neighbors=5
):

    user_id = str(user_id)

    # Check user validity
    if user_id not in user_item_matrix.index:

        print("User ID not found:", user_id)

        print(
            "Cold-start handling: collect user preferences "
            "or use popular-talk recommendations."
        )

        return pd.DataFrame()

    # Get selected user's ratings
    user_ratings = user_item_filled.loc[user_id]

    # Identify talks already rated by the user
    rated_talks = set(
        user_item_matrix.loc[user_id]
        .dropna()
        .index
    )

    # Find similar users
    similarity_scores = (
        user_similarity_df.loc[user_id]
        .drop(user_id)
        .sort_values(ascending=False)
    )

    # Use only users with positive similarity
    similar_users = similarity_scores[
        similarity_scores > 0
    ].head(n_neighbors)

    # Candidate talk recommendations
    recommendations = []

    for talk_id in user_item_matrix.columns:

        # Exclude talks already rated by selected user
        if talk_id in rated_talks:
            continue

        # Ratings given by similar users
        neighbor_ratings = user_item_filled.loc[
            similar_users.index,
            talk_id
        ]

        # Consider only actual ratings
        valid_ratings = neighbor_ratings[
            neighbor_ratings > 0
        ]

        valid_similarities = similar_users.loc[
            valid_ratings.index
        ]

        if len(valid_ratings) == 0:
            continue

        # Similarity-weighted predicted rating
        if valid_similarities.sum() > 0:

            predicted_rating = (
                valid_ratings * valid_similarities
            ).sum() / valid_similarities.sum()

        else:
            continue

        # Get actual talk title
        if talk_id in talk_lookup.index:

            talk_title = talk_lookup.loc[
                talk_id, "title"
            ]

        else:

            talk_title = "Title not available"

        recommendations.append({

            "user_id": user_id,

            "talk_id": talk_id,

            "title": talk_title,

            "predicted_rating": round(
                float(predicted_rating), 3
            ),

            "similar_users_count": len(valid_ratings)

        })

    # --------------------------------------------------------
    # FALLBACK: POPULARITY-BASED RECOMMENDATIONS
    # --------------------------------------------------------

    if not recommendations:

        print(
            "No similar-user recommendations available. "
            "Using average ratings as fallback."
        )

        global_ratings = interactions.groupby(
            "talk_id"
        ).agg(

            predicted_rating=("rating", "mean"),

            rating_count=("rating", "count")

        ).reset_index()

        global_ratings = global_ratings[
            ~global_ratings["talk_id"].isin(rated_talks)
        ]

        global_ratings = global_ratings.sort_values(
            by=["predicted_rating", "rating_count"],
            ascending=False
        )

        for _, row in global_ratings.head(top_n).iterrows():

            talk_id = row["talk_id"]

            title = (
                talk_lookup.loc[talk_id, "title"]
                if talk_id in talk_lookup.index
                else "Title not available"
            )

            recommendations.append({

                "user_id": user_id,

                "talk_id": talk_id,

                "title": title,

                "predicted_rating": round(
                    float(row["predicted_rating"]), 3
                ),

                "similar_users_count": 0

            })

    # Sort by predicted rating
    recommendation_df = pd.DataFrame(
        recommendations
    )

    recommendation_df = recommendation_df.sort_values(
        by="predicted_rating",
        ascending=False
    )

    return recommendation_df.head(
        top_n
    ).reset_index(drop=True)


# ------------------------------------------------------------
# 10. TEST CASE 1: SIMILAR USERS
# ------------------------------------------------------------

print("\nSIMILAR USERS FOR U001:")

display(
    find_similar_users("U001", top_n=5)
)

# ------------------------------------------------------------
# 11. TEST CASE 2: RECOMMENDATIONS FOR THREE USERS
# ------------------------------------------------------------

test_users = (
    interactions.groupby("user_id")
    .size()
    .sort_values(ascending=False)
    .head(3)
    .index.tolist()
)

all_user_recommendations = []

for user_id in test_users:

    print("\n" + "=" * 70)

    print("RECOMMENDATIONS FOR USER:", user_id)

    print("=" * 70)

    print("\nSimilar users:")

    display(
        find_similar_users(
            user_id,
            top_n=5
        )
    )

    print("\nRecommended TED Talks:")

    result = recommend_for_user(
        user_id,
        top_n=5,
        n_neighbors=5
    )

    display(result)

    if not result.empty:

        all_user_recommendations.append(result)

# ------------------------------------------------------------
# 12. TEST INVALID USER
# ------------------------------------------------------------

print("\nINVALID USER TEST:")

invalid_user_result = recommend_for_user(
    "U999",
    top_n=5
)

# ------------------------------------------------------------
# 13. COMBINE RECOMMENDATION RESULTS
# ------------------------------------------------------------

if all_user_recommendations:

    final_results = pd.concat(
        all_user_recommendations,
        ignore_index=True
    )

    print("\nALL USER RECOMMENDATIONS:")

    display(final_results)

    final_results.to_csv(
        "/content/task8_collaborative_recommendations.csv",
        index=False
    )

# ------------------------------------------------------------
# 14. SAVE USER SIMILARITY MATRIX
# ------------------------------------------------------------

user_similarity_df.to_csv(
    "/content/task8_user_similarity_matrix.csv"
)

user_item_matrix.to_csv(
    "/content/task8_user_item_matrix.csv"
)

# ------------------------------------------------------------
# 15. FINAL REPORT
# ------------------------------------------------------------

print("\n" + "=" * 70)

print("TASK 8: COLLABORATIVE FILTERING REPORT")

print("=" * 70)

print("Total users:", user_item_matrix.shape[0])

print("Total TED Talks:", user_item_matrix.shape[1])

print("User similarity matrix:", user_similarity_df.shape)

print("Recommendation function: recommend_for_user()")

print("\nCONCEPTUAL QUESTIONS")

print("""
1. Collaborative Filtering:
   Recommends items based on user interaction patterns.

2. Content-Based vs Collaborative Filtering:
   Content-based uses item content, whereas collaborative
   filtering uses user-item interaction patterns.

3. User-Item Matrix:
   A table where rows represent users and columns represent
   TED Talks. Values represent user ratings.

4. User Similarity:
   Helps identify users who have similar interests.

5. Cosine Similarity:
   Measures similarity between user rating vectors.
   Higher values indicate more similar rating patterns.

6. Excluding Rated Talks:
   Prevents recommending talks the user has already rated.

7. Predicted Rating:
   An estimated rating based on ratings from similar users.

8. New User:
   A new user has no interaction history. Popularity-based
   or content-based recommendations can be used initially.

9. Cold-Start Problem:
   The difficulty of recommending items for new users or
   newly added items without sufficient interaction data.

10. Limitations:
    - Sparse interaction data.
    - Cold-start problem.
    - Popularity bias.
    - Zero-filling can distort similarity.
    - Simulated ratings may not represent real preferences.
""")

print("\nTASK 8 COMPLETED SUCCESSFULLY!")
```
