```python

# ============================================================
# TASK 7: USER PREFERENCES AND INTERACTION DATA
# Notebook: 07_user_interaction_data.ipynb
# ============================================================

import pandas as pd
import numpy as np
from pathlib import Path

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# ------------------------------------------------------------
# 1. LOAD CLEANED TED TALKS DATASET
# ------------------------------------------------------------

DATA_PATH = "/content/ted_talks_cleaned.csv"

if not Path(DATA_PATH).exists():

    from google.colab import files

    print("Upload ted_talks_cleaned.csv")
    files.upload()

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

# ------------------------------------------------------------
# 2. PREPARE TED TALK INFORMATION
# ------------------------------------------------------------

required_columns = ["talk_id", "title", "combined_text"]

for col in required_columns:

    if col not in df.columns:
        raise ValueError(f"Missing column: {col}")

df["combined_text"] = (
    df["combined_text"]
    .fillna("")
    .astype(str)
)

df["title"] = df["title"].fillna("").astype(str)

df = df[
    (df["combined_text"].str.strip() != "") &
    (df["title"].str.strip() != "")
].copy()

df = df.drop_duplicates(
    subset="talk_id"
).reset_index(drop=True)

print("Usable TED Talks:", len(df))

# ------------------------------------------------------------
# 3. GENERATE CONTENT CLUSTERS
# ------------------------------------------------------------

# This is used only to create realistic simulated preferences.
# It does not represent actual user behaviour.

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=5000,
    min_df=2,
    max_df=0.95
)

talk_vectors = vectorizer.fit_transform(
    df["combined_text"]
)

number_of_clusters = min(8, len(df))

kmeans = KMeans(
    n_clusters=number_of_clusters,
    random_state=42,
    n_init=10
)

df["content_cluster"] = kmeans.fit_predict(
    talk_vectors
)

print("\nContent clusters created:", number_of_clusters)

# ------------------------------------------------------------
# 4. SELECT A POOL OF ACTUAL TED TALKS
# ------------------------------------------------------------

talk_pool = df.sample(
    n=min(120, len(df)),
    random_state=42
).copy()

print("TED Talks selected for simulation:", len(talk_pool))

# ------------------------------------------------------------
# 5. SIMULATE USER INTERACTIONS
# ------------------------------------------------------------

rng = np.random.default_rng(42)

number_of_users = 30

interactions = []

# Five users belong to each preference group.
# Each group has two preferred content clusters.

for user_number in range(1, number_of_users + 1):

    user_id = f"U{user_number:03d}"

    preference_group = (user_number - 1) // 5

    preferred_clusters = [
        preference_group % number_of_clusters,
        (preference_group + 1) % number_of_clusters
    ]

    preferred_talks = talk_pool[
        talk_pool["content_cluster"].isin(
            preferred_clusters
        )
    ]

    other_talks = talk_pool[
        ~talk_pool["content_cluster"].isin(
            preferred_clusters
        )
    ]

    # Select talks from preferred and non-preferred clusters
    preferred_sample = preferred_talks.sample(
        n=min(18, len(preferred_talks)),
        random_state=100 + user_number
    )

    remaining_count = 26 - len(preferred_sample)

    other_sample = other_talks.sample(
        n=min(remaining_count, len(other_talks)),
        random_state=200 + user_number
    )

    selected_user_talks = pd.concat([
        preferred_sample,
        other_sample
    ])

    # Fill any remaining records if necessary
    if len(selected_user_talks) < 26:

        remaining_talks = talk_pool[
            ~talk_pool["talk_id"].isin(
                selected_user_talks["talk_id"]
            )
        ]

        extra = remaining_talks.sample(
            n=min(
                26 - len(selected_user_talks),
                len(remaining_talks)
            ),
            random_state=300 + user_number
        )

        selected_user_talks = pd.concat([
            selected_user_talks,
            extra
        ])

    for _, talk in selected_user_talks.iterrows():

        is_preferred = (
            talk["content_cluster"] in preferred_clusters
        )

        if is_preferred:

            rating = int(rng.choice([4, 5], p=[0.35, 0.65]))

        else:

            rating = int(rng.choice([2, 3], p=[0.4, 0.6]))

        # Add small rating variation
        if rng.random() < 0.10:

            rating = int(rng.choice([1, 3, 4, 5]))

        liked = rating >= 4

        disliked = rating <= 2

        saved = liked and (rng.random() < 0.30)

        interactions.append({

            "user_id": user_id,

            "talk_id": talk["talk_id"],

            "rating": rating,

            "liked": liked,

            "disliked": disliked,

            "watched": True,

            "saved": saved,

            "interaction_type": (
                "like" if liked
                else "dislike" if disliked
                else "watched"
            )

        })

# ------------------------------------------------------------
# 6. CREATE INTERACTION DATAFRAME
# ------------------------------------------------------------

interaction_df = pd.DataFrame(interactions)

print("\nINTERACTION DATASET:")
print(interaction_df.shape)

display(interaction_df.head(10))

# ------------------------------------------------------------
# 7. SAVE SIMULATED INTERACTION DATA
# ------------------------------------------------------------

interaction_folder = Path("/content/data/interaction")

interaction_folder.mkdir(
    parents=True,
    exist_ok=True
)

interaction_path = (
    interaction_folder / "user_interactions.csv"
)

interaction_df.to_csv(
    interaction_path,
    index=False
)

print("\nInteraction dataset saved:")
print(interaction_path)

# ------------------------------------------------------------
# 8. CREATE USER-ITEM INTERACTION MATRIX
# ------------------------------------------------------------

user_item_matrix = interaction_df.pivot_table(

    index="user_id",

    columns="talk_id",

    values="rating",

    aggfunc="mean"

)

print("\nUSER-ITEM INTERACTION MATRIX:")
print("Matrix shape:", user_item_matrix.shape)

display(user_item_matrix.head(10))

# ------------------------------------------------------------
# 9. SAVE USER-ITEM MATRIX
# ------------------------------------------------------------

user_item_matrix.to_csv(
    interaction_folder / "user_item_matrix.csv"
)

# ------------------------------------------------------------
# 10. BASIC INTERACTION ANALYSIS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("USER INTERACTION ANALYSIS")
print("=" * 65)

print("Total interactions:", len(interaction_df))

print("Unique users:",
      interaction_df["user_id"].nunique())

print("Unique TED Talks:",
      interaction_df["talk_id"].nunique())

print("Average rating:",
      round(interaction_df["rating"].mean(), 2))

print("Total likes:",
      interaction_df["liked"].sum())

print("Total dislikes:",
      interaction_df["disliked"].sum())

print("Total saved talks:",
      interaction_df["saved"].sum())

# ------------------------------------------------------------
# 11. HIGHEST RATED TED TALKS
# ------------------------------------------------------------

talk_ratings = interaction_df.groupby(
    "talk_id"
).agg(

    average_rating=("rating", "mean"),

    total_ratings=("rating", "count"),

    total_likes=("liked", "sum")

).reset_index()

talk_ratings = talk_ratings.merge(
    df[["talk_id", "title"]],
    on="talk_id",
    how="left"
)

talk_ratings = talk_ratings.sort_values(
    by=["average_rating", "total_ratings"],
    ascending=False
)

print("\nTOP 10 HIGHEST RATED TED TALKS:")

display(talk_ratings.head(10))

# ------------------------------------------------------------
# 12. USERS WITH MOST RATINGS
# ------------------------------------------------------------

user_activity = interaction_df.groupby(
    "user_id"
).agg(

    total_ratings=("rating", "count"),

    average_rating=("rating", "mean"),

    likes=("liked", "sum")

).reset_index()

user_activity = user_activity.sort_values(
    by="total_ratings",
    ascending=False
)

print("\nUSER ACTIVITY:")

display(user_activity.head(10))

# ------------------------------------------------------------
# 13. MOST INTERACTED TED TALKS
# ------------------------------------------------------------

most_interacted = interaction_df.groupby(
    "talk_id"
).size().reset_index(name="interaction_count")

most_interacted = most_interacted.merge(
    df[["talk_id", "title"]],
    on="talk_id",
    how="left"
)

most_interacted = most_interacted.sort_values(
    by="interaction_count",
    ascending=False
)

print("\nMOST INTERACTED TED TALKS:")

display(most_interacted.head(10))

# ------------------------------------------------------------
# 14. VISUALIZE USER INTERACTIONS
# ------------------------------------------------------------

plt.figure(figsize=(10, 5))

sns.countplot(
    data=interaction_df,
    x="rating"
)

plt.title("Distribution of User Ratings")

plt.xlabel("Rating")

plt.ylabel("Number of Interactions")

plt.show()

# ------------------------------------------------------------
# 15. VISUALIZE USER-ITEM MATRIX
# ------------------------------------------------------------

plt.figure(figsize=(15, 8))

sns.heatmap(
    user_item_matrix.iloc[:15, :30],
    cmap="YlGnBu",
    annot=False
)

plt.title("User-Item Interaction Matrix")

plt.xlabel("TED Talk IDs")

plt.ylabel("User IDs")

plt.tight_layout()

plt.show()

# ------------------------------------------------------------
# 16. ANSWERS TO CONCEPTUAL QUESTIONS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("TASK 7 CONCEPTUAL UNDERSTANDING")
print("=" * 65)

print("""
1. User ID:
   Unique identifier assigned to each user.

2. TED Talk ID:
   Unique identifier assigned to each TED Talk.

3. Rating:
   Numerical value representing user satisfaction.

4. Likes and dislikes:
   Binary indicators of user preference.

5. Watch history:
   Records the talks viewed by a user.

6. Saved talks:
   Talks bookmarked for future viewing.

7. User-Item Matrix:
   Rows represent users and columns represent TED Talks.

8. Missing rating:
   Indicates that the user has not rated or interacted
   with that TED Talk in this dataset.

9. Collaborative Filtering:
   Uses patterns in user interactions to recommend
   items preferred by users with similar interests.

10. Simulated Data:
    The user interactions in this notebook are artificially
    generated for educational and model-testing purposes.
    They are not real user activity.
""")

print("\nTASK 7 COMPLETED SUCCESSFULLY!")
```
