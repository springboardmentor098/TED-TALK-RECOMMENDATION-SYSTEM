```python
from urllib.request import urlretrieve
from pathlib import Path

dataset_url = (
    "https://huggingface.co/datasets/"
    "FrancophonIA/TED_talks/resolve/main/ted_talks_en.csv"
)

dataset_path = Path("../data/raw/ted_talks.csv")

dataset_path.parent.mkdir(parents=True, exist_ok=True)

if not dataset_path.exists():
    urlretrieve(dataset_url, dataset_path)
    print("Dataset downloaded successfully!")
else:
    print("Dataset already exists.")

print("Dataset location:", dataset_path.resolve())
```

```python
if "talk_id" in df.columns:
    print("Unique talk IDs:", df["talk_id"].nunique())

    print(
        "Duplicate talk IDs:",
        df["talk_id"].duplicated().sum()
    )
```

```python
if "transcript" in df.columns:

    transcript_available = df["transcript"].notna().sum()

    transcript_missing = df["transcript"].isna().sum()

    print("Available transcripts:", transcript_available)

    print("Missing transcripts:", transcript_missing)

    print(
        "Transcript availability percentage:",
        round(transcript_available / len(df) * 100, 2),
        "%"
    )
```
