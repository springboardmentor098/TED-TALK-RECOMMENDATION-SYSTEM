Distribution and outlier analysis of `views` and `comments`.

**Input:** `/content/processed/03_featured_df.pkl`

## 1. Load featured data

```python
import pandas as pd

clean_df = pd.read_pickle("/content/processed/03_featured_df.pkl")
print(clean_df.shape)
```

## 2. Views & comments: histograms and boxplots

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].hist(clean_df["views"], bins=40)
axes[0, 0].set_title("Views Distribution")

axes[0, 1].boxplot(clean_df["views"])
axes[0, 1].set_title("Views Boxplot")

axes[1, 0].hist(clean_df["comments"], bins=40)
axes[1, 0].set_title("Comments Distribution")

axes[1, 1].boxplot(clean_df["comments"])
axes[1, 1].set_title("Comments Boxplot")

plt.tight_layout()
plt.show()
```
