# 05 - Text Vectorization (TF-IDF)
Build the combined text field and convert it to a TF-IDF matrix - the base for the content-based recommender.

**Input:** `/content/processed/03_featured_df.pkl`  
**Outputs:** `/content/processed/05_final_df.pkl`, `05_tfidf_matrix.npz`, `05_tfidf_vectorizer.joblib`

## 1. Load featured data

```python
import pandas as pd

clean_df = pd.read_pickle("/content/processed/03_featured_df.pkl")
print(clean_df.shape)
```

## 2. Combine text fields (title + description + tags + transcript)

```python
clean_df["combined_text"] = (
    clean_df["title"].fillna("") + " " +
    clean_df["description"].fillna("") + " " +
    clean_df["tags_text"].fillna("") + " " +
    clean_df["transcript"].fillna("")
).str.strip()

print(clean_df[
    ["title", "tags_text", "combined_text"]
].head(2))
```

## 3. TF-IDF vectorization

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(
    max_features=10000,
    stop_words="english",
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

X_text = tfidf.fit_transform(
    clean_df["combined_text"]
)

print("Number of talks:", X_text.shape[0])
print("Number of text features:", X_text.shape[1])
print("Matrix shape:", X_text.shape)
```

## 4. Save outputs

```python
import joblib
from scipy import sparse

clean_df.to_pickle("/content/processed/05_final_df.pkl")
sparse.save_npz("/content/processed/05_tfidf_matrix.npz", X_text)
joblib.dump(tfidf, "/content/processed/05_tfidf_vectorizer.joblib")
print("Saved final dataframe, TF-IDF matrix and vectorizer")
```
