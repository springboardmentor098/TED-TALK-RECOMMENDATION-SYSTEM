## 1. Import libraries

```python
import pandas as pd
```

```python
import os
os.makedirs("/content/processed", exist_ok=True)
```

## 2. Load TED talks metadata

```python
ted_talks_dataset=pd.read_csv('/content/ted_main.csv')
```

```python
ted_talks_dataset
```

```python
ted_talks_dataset.shape
```

```python
ted_talks_dataset.info()
```

## 3. Load transcripts

```python
transcripts_dataset=pd.read_csv('/content/transcripts.csv')
```

```python
transcripts_dataset
```

```python
transcripts_dataset.shape
```

## 4. Merge both datasets on `url`

```python
merged_dataset = pd.merge(ted_talks_dataset, transcripts_dataset, on='url', how='left')
print(f"Merged shape: {merged_dataset.shape}")
display(merged_dataset.head())
```

## 5. Save output for the next notebook

```python
merged_dataset.to_pickle("/content/processed/01_merged_dataset.pkl")
print("Saved:", merged_dataset.shape)
```
