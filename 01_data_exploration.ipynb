```python

# ============================================================
# TED TALKS RECOMMENDATION SYSTEM
# TASK 1: DATA LOADING AND EXPLORATORY DATA ANALYSIS (EDA)
# ============================================================

# 1. Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_colwidth", 100)
sns.set_theme(style="whitegrid")

print("=" * 60)
print("TED TALKS RECOMMENDATION SYSTEM")
print("TASK 1: DATA EXPLORATION")
print("=" * 60)

# 2. Locate and load the dataset
dataset_path = Path("../data/raw/ted_talks.csv")

if not dataset_path.exists():
    dataset_path = Path("data/raw/ted_talks.csv")

if not dataset_path.exists():
    raise FileNotFoundError(
        "Dataset not found. Check that ted_talks.csv "
        "is inside the data/raw folder."
    )

df = pd.read_csv(dataset_path)

print("\n1. DATASET LOADED SUCCESSFULLY")
print("Dataset path:", dataset_path.resolve())

# 3. Display first five records
print("\n2. FIRST FIVE RECORDS")
display(df.head())

# 4. Display last five records
print("\n3. LAST FIVE RECORDS")
display(df.tail())

# 5. Dataset dimensions
print("\n4. DATASET DIMENSIONS")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

# 6. Column names
print("\n5. COLUMN NAMES")
for column in df.columns:
    print("-", column)

# 7. Dataset information
print("\n6. DATASET INFORMATION")
df.info()

# 8. Statistical summary
print("\n7. NUMERICAL STATISTICAL SUMMARY")
display(df.describe())

print("\n8. CATEGORICAL STATISTICAL SUMMARY")
display(df.describe(include="object").T)

# 9. Missing values
print("\n9. MISSING VALUES")
missing_values = df.isnull().sum()
missing_summary = pd.DataFrame({
    "Missing Count": missing_values,
    "Missing Percentage": (
        missing_values / len(df) * 100
    ).round(2)
})
display(
    missing_summary.sort_values(
        by="Missing Count", ascending=False
    )
)

# 10. Duplicate records
print("\n10. DUPLICATE RECORDS")
duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)

# 11. Unique values in each column
print("\n11. UNIQUE VALUES IN EACH COLUMN")
for column in df.columns:
    print(f"{column}: {df[column].nunique(dropna=True)}")

# 12. Check important text columns
print("\n12. CHECK IMPORTANT TED TALK COLUMNS")
for column in ["title", "description", "transcript", "speaker"]:
    if column in df.columns:
        print(f"\n{column.upper()}")
        display(df[column].dropna().head(3))

# 13. Plot missing values
print("\n13. MISSING VALUES VISUALIZATION")
missing_plot = missing_values[missing_values > 0].sort_values(
    ascending=False
)

if not missing_plot.empty:
    plt.figure(figsize=(12, 5))
    missing_plot.plot(kind="bar")
    plt.title("Missing Values by Column")
    plt.xlabel("Columns")
    plt.ylabel("Number of Missing Values")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()
else:
    print("No missing values found.")

# 14. Plot numerical column distributions
print("\n14. NUMERICAL DATA DISTRIBUTIONS")
numeric_columns = df.select_dtypes(include=np.number).columns.tolist()

if numeric_columns:
    df[numeric_columns].hist(
        figsize=(14, 10),
        bins=20,
        edgecolor="black"
    )
    plt.suptitle("Distribution of Numerical Features")
    plt.tight_layout()
    plt.show()
else:
    print("No numerical columns available for plotting.")

# 16. Dataset exploration summary
print("\n" + "=" * 60)
print("16. EXPLORATION SUMMARY")
print("=" * 60)
print("Total TED Talk records:", len(df))
print("Total columns:", len(df.columns))
print("Duplicate rows:", duplicate_count)
print("Columns with missing values:", int((missing_values > 0).sum()))

print("\nDATA EXPLORATION COMPLETED SUCCESSFULLY!")
print("Note: Missing values and duplicates have been inspected,")
print("but the original dataset has not been modified.")
```

```python

# Task 2 - Activity 2: Missing Values and Duplicates

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("../data/raw/ted_talks.csv")

# 1. Missing values and percentages
missing_count = df.isnull().sum()
missing_percentage = (missing_count / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage (%)": missing_percentage.round(2)
})

print("===== MISSING VALUE REPORT =====")
display(missing_report)

# 2. Duplicate analysis
print("\n===== DUPLICATE ANALYSIS =====")
print("Duplicate complete rows:", df.duplicated().sum())
print("Duplicate TED Talk IDs:", df["talk_id"].duplicated().sum())

# 3. Check missing and empty important fields
print("\n===== IMPORTANT FIELD CHECK =====")

important_columns = ["title", "description", "topics", "transcript", "url"]

for column in important_columns:
    if column in df.columns:
        values = df[column].fillna("").astype(str).str.strip()
        print(f"\n{column}")
        print("Missing values:", df[column].isnull().sum())
        print("Empty values:", values.eq("").sum())
    else:
        print(f"\n{column}: Column not available")
```

```python

# Task 2 - Activity 3: Text Analysis

for column in ["description", "transcript"]:
    text_data = df[column].fillna("").astype(str).str.strip()

    df[f"{column}_char_count"] = text_data.str.len()
    df[f"{column}_word_count"] = text_data.apply(
        lambda text: len(text.split())
    )

    print(f"\n===== {column.upper()} ANALYSIS =====")
    print("Empty entries:", text_data.eq("").sum())
    print("Entries with fewer than 10 words:",
          (text_data.str.split().str.len() < 10).sum())

    print("\nCharacter count statistics:")
    display(df[f"{column}_char_count"].describe())

    print("Word count statistics:")
    display(df[f"{column}_word_count"].describe())

# Required visualization: transcript length distribution
plt.figure(figsize=(10, 5))
sns.histplot(df["transcript_word_count"], bins=30)
plt.title("Distribution of TED Talk Transcript Word Counts")
plt.xlabel("Transcript Word Count")
plt.ylabel("Number of TED Talks")
plt.tight_layout()
plt.show()
```

```python

# Task 2 - Activity 4: Tag and Topic Analysis

from collections import Counter
import ast

topic_counter = Counter()

for value in df["topics"].dropna():
    if isinstance(value, str):
        try:
            parsed_topics = ast.literal_eval(value)
            if isinstance(parsed_topics, list):
                topics_list = parsed_topics
            else:
                topics_list = [value]
        except (ValueError, SyntaxError):
            topics_list = [topic.strip() for topic in value.split(",")]
    elif isinstance(value, list):
        topics_list = value
    else:
        continue

    for topic in topics_list:
        if isinstance(topic, str) and topic.strip():
            topic_counter[topic.strip()] += 1

top_topics = topic_counter.most_common(15)

topic_report = pd.DataFrame(
    top_topics, columns=["Topic", "Frequency"]
)

print("Total distinct topics:", len(topic_counter))
print("\nTop 15 most frequent topics:")
display(topic_report)

plt.figure(figsize=(10, 6))
sns.barplot(data=topic_report, x="Frequency", y="Topic")
plt.title("Top 15 Most Frequent TED Talk Topics")
plt.xlabel("Frequency")
plt.ylabel("Topic")
plt.tight_layout()
plt.show()
```
