```python
# Keep only the columns needed for recommendations
required_columns = [
    "talk_id", "title", "speaker_1", "views",
    "topics", "description", "transcript"
]

df_clean = df[required_columns].copy()

# Remove rows without a title or transcript
df_clean = df_clean.dropna(subset=["title", "transcript"])

# Remove duplicate talk IDs
df_clean = df_clean.drop_duplicates(subset=["talk_id"])

# Reset row numbering
df_clean = df_clean.reset_index(drop=True)

print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", df_clean.shape)
print("Remaining missing values:")
print(df_clean.isnull().sum())
```

```python
import os

output_folder = r"C:\Users\anish\ted-talks-recommender\data\processed"
os.makedirs(output_folder, exist_ok=True)

output_path = os.path.join(output_folder, "ted_talks_cleaned.csv")
df_clean.to_csv(output_path, index=False)

print("Cleaned dataset saved successfully!")
print(output_path)
```

# Dataset Inspection Report

## 1. Dataset Name and Source

- **Dataset Name:** TED Talks Transcripts for NLP
- **Source:** Kaggle
- **Original File:** `ted_talks_en.csv`
- **Dataset Format:** CSV (Comma-Separated Values)

## 2. Dataset Dimensions

- **Total Rows:** 4,005
- **Total Columns:** 19

The dataset contains information about TED Talks, including titles, speakers, topics, descriptions, views, and transcripts.

## 3. Unique TED Talks

- **Unique Talk IDs:** 4,005
- **Duplicate Talk IDs:** 0
- **Duplicate Rows:** 0

All 4,005 records have unique talk IDs, and no completely duplicated rows were found.

## 4. Important Columns

The dataset contains the following important columns:

- `talk_id`: Unique identifier for each TED Talk.
- `title`: Title of the TED Talk.
- `speaker_1`: Name of the main speaker.
- `description`: Description or summary of the talk.
- `topics`: Topics related to the talk.
- `transcript`: Spoken text of the TED Talk.
- `url`: Link to the original TED Talk.
- `views`: Number of views.
- `duration`: Duration of the talk.
- `event`: Name of the event where the talk was presented.
- `published_date`: Date when the talk was published.

## 5. Data Quality and Missing Values

The dataset contains missing values in the following columns:

| Column | Missing Values |
|---|---:|
| `all_speakers` | 4 |
| `occupations` | 522 |
| `about_speakers` | 503 |
| `recorded_date` | 1 |
| `comments` | 655 |

The important columns `title`, `speaker_1`, `description`, `topics`, and `transcript` have no missing values.

No duplicate talk IDs or duplicate rows were found. The original dataset has not been cleaned or modified during this inspection.

## 6. Transcript Availability

- **Total Transcripts:** 4,005
- **Missing Transcripts:** 0
- **Empty Transcripts:** 0
- **Available Transcripts:** 4,005
- **Transcript Availability:** 100%

All records contain non-empty transcripts, which is useful for text-based recommendation.

## 7. Dataset Limitations

The dataset has some limitations:

- Speaker occupation details are missing for 522 records.
- Speaker biography details are missing for 503 records.
- Comments are missing for 655 records.
- One record has a missing `recorded_date`.
- Topic labels and transcript text may require further preprocessing before building the recommendation system.
- The dataset source and usage licence should be verified before redistribution or public release.

## 8. Suitability for Recommendation

The dataset is suitable for developing a content-based TED Talks recommendation prototype.

The `title`, `description`, `topics`, and `transcript` columns provide useful text for comparing talks and recommending similar content. Since all 4,005 records have available transcripts and unique talk IDs, the dataset provides a useful starting point for the project.

Further exploratory data analysis, text preprocessing, and feature extraction will be performed in the next stage.

## 9. Conclusion

The dataset contains 4,005 TED Talks and 19 columns. It has no duplicate talk IDs or duplicate rows, and all transcripts are available. Although some optional metadata contains missing values, the main text fields needed for a content-based recommendation prototype are complete.

Therefore, the dataset is a promising starting point for the project, subject to verification of its licence and suitability for the intended use.
