# 02 - Data Cleaning
Missing values, duplicates and text whitespace cleanup.

**Input:** `/content/processed/01_merged_dataset.pkl`  
**Output:** `/content/processed/02_clean_df.pkl`

Next notebook: `03_Feature_Engineering.ipynb`

## 1. Load merged dataset

```python
import pandas as pd

merged_dataset = pd.read_pickle("/content/processed/01_merged_dataset.pkl")
print(merged_dataset.shape)
```

## 2. Handle missing values

```python
merged_dataset.isnull().sum()
```

```python
# Fill missing values in 'transcript' and 'speaker_occupation' using fillna
merged_dataset['transcript'] = merged_dataset['transcript'].fillna('No transcript available')
merged_dataset['speaker_occupation'] = merged_dataset['speaker_occupation'].fillna('Unknown')
```

```python
 # Verify that there are no remaining missing values in these columns
print("Missing values after filling:")
print(merged_dataset[['transcript', 'speaker_occupation']].isnull().sum())
```

```python
# after data cleaning

merged_dataset.isnull().sum()
```

## 3. Remove duplicates

```python
merged_dataset.duplicated().sum()
```

```python
# Remove duplicate rows, keeping only the first occurrence
merged_dataset = merged_dataset.drop_duplicates(keep='first')

# Verify the results
print(f"New shape of merged_dataset: {merged_dataset.shape}")
print(f"Remaining duplicates: {merged_dataset.duplicated().sum()}")
```

```python
merged_dataset.duplicated().sum()
```

## 4. Keep a working copy

```python
# keep the original dataset

clean_df = merged_dataset.copy()
```

## 5. Clean text columns (whitespace / newlines)

```python
# This removes repeated whitespace and newline characters.
# It does not remove meaningful text or automatically lowercase all columns.


text_columns = [
    "title",
    "description",
    "transcript",
    "main_speaker",
    "speaker_occupation",
    "event",
    "name",
    "url"
]

for col in text_columns:
    clean_df[col] = (
        clean_df[col]
        .astype("string")
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )
```

```python
# Check whether the URLs are clean
print(
    "URLs with newline characters:",
    clean_df["url"].str.contains(
        r"[\r\n]", regex=True, na=False
    ).sum()
)
```

## 6. Save output for the next notebook

```python
clean_df.to_pickle("/content/processed/02_clean_df.pkl")
print("Saved:", clean_df.shape)
```
